Contiene datos de los clientes que realizaron compras en la tienda virtual, siendo estos: ID de la compra, ID del cliente, la ciudad, estado y región de la compra, edad, sexo biológico, si participan en el programa de cashback y la calificación, o nota, de la compra.

No es necesario utilizar "print()" para mostrar el resultado de las consultas.

```python
# 1. Información general del DataFrame
print("\n[1] Información general del DataFrame:")
clientes.info()

# 2. Primeras filas para una visión inicial de los datos
print("\n[2] Primeras 5 filas del DataFrame:")
display(clientes.head())

# 3. Dimensiones del DataFrame (filas, columnas)
print("\n[3] Dimensiones del DataFrame (filas, columnas):")
print(clientes.shape)

# 4. Tipos de datos por columna
print("\n[4] Tipos de datos por columna:")
print(clientes.dtypes)

# 5. Estadísticas descriptivas para columnas numéricas
print("\n[5] Estadísticas descriptivas para columnas numéricas:")
display(clientes.describe())

# 6. Estadísticas para columnas categóricas
print("\n[6] Estadísticas para columnas categóricas:")
display(clientes.describe(include='object'))

# 7. Distribución de clientes por sexo
print("\n[7] Distribución de clientes por sexo:")
print(clientes['sexo_biologico'].value_counts())

# 8. Participación en el programa de cashback
print("\n[8] Participación en el programa de cashback:")
print(clientes['cashback'].value_counts())

# 9. Distribución de clientes por región
print("\n[9] Distribución de clientes por región:")
print(clientes['region'].value_counts())

# 10. Distribución de clientes por estado
print("\n[10] Distribución de clientes por estado:")
print(clientes['estado'].value_counts())

# 11. Número de ciudades únicas
print("\n[11] Número de ciudades únicas:")
print(clientes['ciudad'].nunique())

# 12. Verificación de valores nulos por columna
print("\n[12] Verificación de valores nulos por columna:")
print(clientes.isnull().sum())

# 13. Histograma de la edad
print("\n[13] Histograma de la edad de los clientes:")
clientes['edad'].hist(bins=20, figsize=(8,5), grid=False)

# 14. Nota promedio por sexo
print("\n[14] Nota promedio por sexo:")
print(clientes.groupby('sexo_biologico')['nota'].mean())

# 15. Nota promedio por región
print("\n[15] Nota promedio por región:")
print(clientes.groupby('region')['nota'].mean())

# 16. Nota promedio general de las compras
print("\n[16] Nota promedio general de todas las compras:")
print(round(clientes['nota'].mean(), 2))

# 17. Número de compras por cliente
print("\n[17] Número de compras por cliente (top 10):")
compras_por_cliente = clientes['ID_cliente'].value_counts()
print(compras_por_cliente.head(10))
```

### **Base de ventas**

> #### <strong><font color="7b71ff" size =6>Prompt:</font></strong>

Ahora vamos explorar la base de ventas en formato de un DataFrame, llamado "ventas". Contiene datos de las ventas del comercio electrónico, que son: ID de la compra, fecha de la compra, hora de la compra, categoría del producto, precio unitario del producto, cantidad vendida, envío y el método de pago.

Queremos hacer una breve exploración de los datos para comprender un poco sobre ellos. Recordando que los datos de esta base también están limpios y tratados. Construye el código para ejecutar esta etapa utilizando Python con el objetivo de conocer mejor el dataframe "clientes" y algunos de sus atributos.

```python
# 1. Información general del DataFrame
print("\n[1] Información general del DataFrame 'ventas':")
ventas.info()

# 2. Primeras 5 filas del DataFrame
print("\n[2] Primeras 5 filas del DataFrame:")
display(ventas.head())

# 3. Dimensiones del DataFrame (filas, columnas)
print("\n[3] Dimensiones del DataFrame:")
print(ventas.shape)

# 4. Tipos de datos por columna
print("\n[4] Tipos de datos por columna:")
print(ventas.dtypes)

# 5. Estadísticas descriptivas para columnas numéricas
print("\n[5] Estadísticas descriptivas (columnas numéricas):")
display(ventas.describe())

# 6. Estadísticas para columnas categóricas
print("\n[6] Estadísticas descriptivas (columnas categóricas):")
display(ventas.describe(include='object'))

# 7. Métodos de pago disponibles
print("\n[7] Métodos de pago disponibles:")
print(ventas['metodo_pago'].value_counts())

# 8. Categorías de productos más frecuentes
print("\n[8] Categorías de productos más frecuentes:")
print(ventas['categoria'].value_counts())

# 9. Verificación de valores nulos por columna
print("\n[9] Verificación de valores nulos por columna:")
print(ventas.isnull().sum())

# 10. Precio unitario promedio
print("\n[10] Precio unitario promedio:")
print(round(ventas['precio_unitario'].mean(), 2))

# 11. Cantidad total de productos vendidos
print("\n[11] Cantidad total de productos vendidos:")
print(ventas['cantidad'].sum())

# 12. Crear columna 'valor_venta' = precio_unitario * cantidad
print("\n[12] Crear nueva columna 'valor_venta':")
ventas['valor_venta'] = ventas['precio_unitario'] * ventas['cantidad']
display(ventas[['ID_compra', 'categoria', 'valor_venta']].head())

# 13. Ingreso total del e-commerce en 2023
print("\n[13] Ingreso total del e-commerce en 2023:")
print(round(ventas['valor_venta'].sum(), 2))

# 14. Costo total de envíos
print("\n[14] Costo total de envíos:")
print(round(ventas['envio'].sum(), 2))

# 15. Ingreso total por categoría
print("\n[15] Ingreso total por categoría de producto:")
print(ventas.groupby('categoria')['valor_venta'].sum().sort_values(ascending=False))

# 16. Ingreso total por método de pago
print("\n[16] Ingreso total por método de pago:")
print(ventas.groupby('metodo_pago')['valor_venta'].sum().sort_values(ascending=False))
```

Ahora vamos a unir las dos bases de nuestro proyecto en una sola utilizando como llave la columna "ID_compra". Además, organizaremos las columnas en la siguiente secuencia:

* ID_compra

* horario
* categoria

* cantidad
* envio
* metodo_pago
* ID_cliente
* edad

Construye un código para ejecutar esta etapa usando Python y la biblioteca Pandas.


```python
# Los Dataframe clientes y ventas ya estan cargados

# 1. Unir los DataFrames 'ventas' y 'clientes' por la columna 'ID_compra'
print("\n[1] Uniendo DataFrames por 'ID_compra'...")
df = pd.merge(ventas, clientes, on='ID_compra', how='inner')

# 2. Verificar cantidad de registros después de la unión
print("\n[2] Dimensiones del DataFrame combinado:")
print(df.shape)

# 3. Reorganizar columnas en el orden especificado
orden_columnas = [
    'ID_compra', 'fecha', 'horario', 'categoria',
    'precio_unitario', 'cantidad', 'envio', 'metodo_pago',
    'ID_cliente', 'edad', 'sexo_biologico', 'ciudad',
    'estado', 'region', 'cashback', 'nota'
]

df = df[orden_columnas]

# 4. Mostrar las primeras filas del DataFrame final
print("\n[3] Primeras filas del DataFrame combinado y ordenado:")
display(df.head())
```

## ✅ Sugerencias de Visualizaciones de Zoop

Para construir una narrativa sólida a partir de los datos de ventas y clientes de Zoop, se pueden realizar varias visualizaciones que te ayuden a explorar la facturación, el perfil de los clientes, las relaciones entre variables y las tendencias de comportamiento. A continuación te presento las visualizaciones agrupadas por enfoque analítico:

---

### 1. Análisis de Facturación a lo Largo del Año:

* **📈 Gráfico de líneas de facturación mensual:**
  Permite visualizar cómo evoluciona la facturación mes a mes durante el año. Se puede usar la columna `fecha` agrupada por mes y calcular el ingreso total (`precio_unitario * cantidad + envio`). Esto ayuda a identificar patrones estacionales o meses con menor rendimiento.

* **📊 Gráfico de barras por categoría:**
  Muestra la facturación generada por cada categoría de producto. Es útil para saber cuáles son los productos más rentables y tomar decisiones sobre inventario, promoción y foco de ventas.

* **🗓️🔥 Mapa de calor de días y horas de mayor facturación:**
  Utilizando las columnas `fecha` y `horario`, este gráfico muestra en qué días de la semana y en qué horarios se generan más ventas. Ayuda a definir estrategias de marketing según los momentos de mayor actividad.

---

### 2. Perfil del Cliente:

* **📊 Histograma de edades:**
  Visualiza la distribución de edades de los clientes. Te permite identificar qué grupos etarios son más frecuentes y enfocar estrategias de comunicación y diseño de productos para ellos.

* **⚖️ Gráfico circular o de barras del sexo biológico:**
  Muestra la proporción de clientes masculinos y femeninos (`sexo_biologico`). Es útil para comprender el perfil demográfico de los compradores.

* **📍 Gráfico de barras por región:**
  Permite ver cuántos clientes provienen de cada región del país (`region`). Así puedes entender mejor la distribución geográfica de tu mercado y decidir dónde enfocar campañas regionales.

* **💸 Gráfico de barras por participación en el programa de cashback:**
  Analiza cuántos clientes participan en el programa de cashback (`cashback`). También puedes cruzar esta variable con la `nota` para evaluar su efecto sobre la satisfacción.

---

### 3. Relaciones entre Datos:

* **🔘 Gráfico de dispersión (scatterplot) de precio unitario vs cantidad:**
  Permite observar si hay una relación entre el precio de los productos y la cantidad que se vende. Ayuda a detectar productos de alto volumen o premium.

* **📦 Gráfico de cajas (boxplot) de las notas por método de pago:**
  Muestra si hay diferencias en la evaluación de la compra según el método de pago (`metodo_pago`). Es útil para descubrir problemas en ciertas pasarelas de pago.

* **🧾 Gráfico de barras apiladas de método de pago por región:**
  Visualiza la preferencia de métodos de pago en distintas regiones. Ayuda a adaptar las opciones de pago a las preferencias locales.

---

### 4. Análisis Temporal y de Ciclos:

* **📅 Gráfico de barras por día de la semana:**
  Revela en qué días se realizan más ventas. Puede servir para lanzar ofertas específicas según el comportamiento semanal de los clientes.

* **📊 Gráfico de barras de la evaluación promedio por mes:**
  Permite analizar cómo cambia la satisfacción de los clientes a lo largo del año (evaluación promedio o `nota`). Es útil para detectar periodos donde la experiencia puede mejorar.

---

### 5. Análisis de Métodos de Pago:

* **🥧 Gráfico de torta del método de pago:**
  Muestra la distribución del uso de métodos de pago por parte de los clientes. Permite detectar preferencias generales y tomar decisiones sobre alianzas o promociones específicas.

* **📊 Gráfico de barras de la facturación total por método de pago:**
  Revela cuál método de pago genera más ingresos. Esta información puede usarse para negociar tarifas con plataformas de pago o enfocar campañas según la rentabilidad.

---

### 6. Segmentación del Cliente:

* **👥 Gráfico de barras de cantidad de compras por cliente:**
  Muestra cuántos clientes son compradores frecuentes y cuántos solo compraron una vez. Ayuda a identificar oportunidades para programas de fidelización o campañas de recompra.

* **📊 Análisis de correlación con mapa de calor:**
  Usando un heatmap de correlaciones, puedes ver relaciones entre variables como edad, nota, cantidad comprada, método de pago, cashback, etc. Esto ayuda a descubrir patrones o perfiles de cliente valiosos.

---

Estas visualizaciones te permitirán entender mejor el comportamiento de los clientes, identificar oportunidades comerciales y tomar decisiones estratégicas más informadas para mejorar la experiencia de compra y aumentar la rentabilidad de Zoop.
