## PROMPT

Tenemos un DataFrame llamado "df" y queremos construir un gráfico de barras horizontales que muestre la facturación por region de Zoop.

Deseamos crear una visualización que muestre la facturación de cada region, donde la facturación se calcula multiplicando el precio unitario por la cantidad y sumando el costo de envío.

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
#✅ Paso 1: Calcular la columna de facturación

# Crear la columna 'facturacion' en el DataFrame
df['facturacion'] = (df['precio_unitario'] * df['cantidad']) + df['envio']
````

````python
#✅ Paso 2: Agrupar facturación por región

# Agrupar por región y sumar la facturación
facturacion_region = df.groupby('region')['facturacion'].sum().reset_index()

# Ordenar para visualización más clara
facturacion_region = facturacion_region.sort_values(by='facturacion', ascending=True)
````

````python
#✅ Paso 3: Crear gráfico de barras horizontales

import matplotlib.pyplot as plt
import seaborn as sns

# Estilo visual
sns.set(style="whitegrid")

# Crear gráfico
plt.figure(figsize=(10, 6))
sns.barplot(data=facturacion_region, x='facturacion', y='region', palette='Blues_d')

# Títulos y etiquetas
plt.title('Facturación por Región - Zoop', fontsize=14)
plt.xlabel('Facturación Total')
plt.ylabel('Región')
plt.tight_layout()

# Mostrar gráfico
plt.show()
````
